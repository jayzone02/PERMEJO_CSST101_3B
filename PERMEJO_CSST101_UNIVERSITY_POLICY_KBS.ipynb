Table 1. Debate Arguments

```r
debate_arguments <- data.frame(
  Argument_ID = c("A001", "A002", "A003", "A004", "A005", "A006"),
  Position = c("For", "For", "For", "Against", "Against", "Against"),
  Topic = c("Learning", "Productivity", "AI Literacy", "Academic Integrity", "Dependency", "Assessment"),
  Claim = c(
    "AI can provide students with immediate learning support.",
    "AI can help students complete routine academic tasks more efficiently.",
    "Allowing AI use can help students develop AI literacy.",
    "Undisclosed AI-generated work may create academic integrity concerns.",
    "Excessive AI use may reduce students' independent problem-solving practice.",
    "AI use may make it more difficult to determine whether submitted work reflects student learning."
  ),
  stringsAsFactors = FALSE
)

(debate_arguments)
```

Table 2. Count The Arguments

```r
# Create the evidence data frame[cite: 2]
evidence_table <- data.frame(
  Evidence_ID = c("E001", "E002", "E003", "E004", "E005", "E006"),
  Argument_ID = c("A001", "A002", "A003", "A004", "A005", "A006"),
  Evidence_Type = c("Research", "Research", "Policy", "Policy", "Research", "Research"),
  Strength = c("Strong", "Strong", "Moderate", "Strong", "Strong", "Moderate"),
  stringsAsFactors = FALSE
)

# Display using simple print
(evidence_table)
```

3. Retrieve Arguments Supporting the Polic

```r
for_arguments <- subset(
  debate_arguments,
  Position == "For"
)
(for_arguments)
```

4. Retrieve Arguments Against the Policy

```r
against_arguments <- subset(
  debate_arguments,
  Position == "Against"
)
(against_arguments)
```

5. Connect Arguments and Evidence

```r
knowledge_base <- merge(debate_arguments, evidence_table, by = "Argument_ID")

(knowledge_base)
```

6. Convert Evidence Strength into a Score

```r
get_score <- function(strength){
  if (strength == "Strong"){
    return(3)
  } else if (strength == "Moderate"){
    return(2)
  } else if (strength == "Weak"){
    return(1)
  } else{
    return(0)
  }
}

get_score("Strong")
```

7. Add Evidence Score

```r
knowledge_base$Evidence_Score <- sapply(
  knowledge_base$Strength,
  get_score


)
(knowledge_base)
```

8. Classify Evidence Support

```r
classify_support <- function(score){

  if (score == 3){
    return("Strongly Supported")
  } else if (score == 2){
    return("Moderately Supported")
  } else if (score == 1){
    return("Weakly Supported")
  } else{
    return("No Support")
  }
}

classify_support(2)
```

9. Search by Topic

```r
search_topic<- function(topic){
  result <- subset(
  knowledge_base,
  Topic == topic
  )

  if(nrow(result) == 0)(
    cat("Topic not found.\n")
  )else(
    print(result)
  )
}
```

```r
search_topic("Learning")
```
