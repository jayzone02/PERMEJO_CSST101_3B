12. ADD MORE RULES

```r
students <- data.frame(
  Student_ID = c("S001", "S002", "S003", "S004", "S005"),
  Attendance = c(95, 72, 88, 79, 68),
  Grade = c(90, 70, 85, 77, 65)
)

student_risk <- function(attendance, grade) {
  if (attendance < 75 & grade < 75) {
    return("High Risk")
  } else if (attendance < 85 | grade < 80) {
    return("Moderate Risk")
  } else if (attendance >= 85 & grade >= 80) {
    return("Low Risk")
  }
}

students$Risk <- mapply(student_risk, students$Attendance, students$Grade)

print(format(students, justify = "centre"), row.names = FALSE)
```

13. APPLY THE RULE TO MULTIPLE STUDENT

16. ADD A RECOMMENDATION

```r
get_recommendation <- function(risk) {
  if (risk == "High Risk") {
    return("Immediate academic intervention")
  } else if (risk == "Moderate Risk") {
    return("Academic monitoring")
  } else if (risk == "Low Risk") {
    return("Continue regular monitoring")
  }
}

students$Recommendation <- sapply(students$Risk, get_recommendation)

students
```

17. MINI KRR SYSTEM

```r
mini_krr_system <- function(student_id, attendance, grade) {

  if (attendance < 75 & grade < 75) {
    risk <- "High Risk"
    recommendation <- "Immediate academic intervention"
  } else if (attendance < 85 | grade < 80) {
    risk <- "Moderate Risk"
    recommendation <- "Academic monitoring"
  } else {
    risk <- "Low Risk"
    recommendation <- "Continue regular monitoring"
  }

  cat("==========================================\n")
  cat("         MINI KRR SYSTEM OUTPUT           \n")
  cat("==========================================\n")
  cat("Student ID    :", student_id, "\n")
  cat("Attendance    :", attendance, "%\n")
  cat("Grade         :", grade, "\n")
  cat("Classification:", risk, "\n")
  cat("Recommendation:", recommendation, "\n")
  cat("==========================================\n")
}

students <- data.frame(
  Student_ID = c("S001", "S002", "S003", "S004", "S005"),
  Attendance = c(95, 72, 88, 79, 68),
  Grade = c(90, 70, 85, 77, 65)
)

students$Risk <- mapply(
  function(a, g) {
    if (a < 75 & g < 75) "High Risk"
    else if (a < 85 | g < 80) "Moderate Risk"
    else "Low Risk"
  },
  students$Attendance,
  students$Grade
)

students$Recommendation <- mapply(
  function(a, g) {
    if (a < 75 & g < 75) "Immediate academic intervention"
    else if (a < 85 | g < 80) "Academic monitoring"
    else "Continue regular monitoring"
  },
  students$Attendance,
  students$Grade
)

print(format(students, justify = "centre"), row.names = FALSE)

mini_krr_system("S002", 72, 70)
```

 **Academic Risk:** High Risk

 **Intervention Required:** Immediate academic intervention
